#### 1) What is Prompt?
A simple input instruction we give to an LLM.
If the prompts are hardcoded, there is no structure, scalability and control over them. So we need Prompt Template.

#### 2) Prompt Template.
A Prompt Template is a reusable prompt with placeholders that get filled at runtime.
Think of it like an f-string in python.

#### A) Basic (non-chat model) Prompt Template

```python
# Step 1) Creating a simple PromptTemplate

from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    template= "Explain {topic} in simple terms", # Here {topic} is a placeholder
    input_variables= ['topic']
)
```

```python
# Step 2) Filling in the Template

prompt.format(topic = 'Machine Learning')
```

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model='gpt-4o-mini', temperature=0)
```

```python
# Step 3) Connecting template to a Model
# This is a runnable pipeline

chain = prompt | model
```

```python
# Step 4) Calling the Model

response = chain.invoke({'topic': 'Machine Learning'})
print(response.content)
```

#### B) ChatPromptTemplate
Creates structured chat messages, not raw texts.

```python
## Load the Model

from langchain_openai import ChatOpenAI

model = ChatOpenAI(model='gpt-4o-mini', temperature=0)
```

```python
from langchain_core.prompts import ChatPromptTemplate

chatprompt = ChatPromptTemplate.from_messages(
    [
        ('system', 'You are a helpful AI teacher.'),
        ('human', "Explain {topic} in simple terms.")
    ]
)

chatprompt.format_messages(topic = 'RAG') # it just fills the variables
```

```python
chatchain = chatprompt | model

chatresponse = chatchain.invoke({'topic': 'RAG in Machine Learning'})
print(chatresponse.content)
```

#### 3) RAG-style Prompt template

```python
## Creating prompt

prompt = ChatPromptTemplate([
    ('system', 'Answer ONLY using the given context.'),
    ('human', """
Context:
{context}
     
Question:
{question}
""")
])
```

```python
## Giving inputs

inputs = {
    "context" : "RAG stands for Retrieval-Augmented Generation. It combines retrieval with generation.",
    'question': 'What is RAG'
}
```

```python
## Running the model

chain = prompt | model

response = chain.invoke(inputs)
print(response.content)
```

```python
response1 = chain.invoke({'context': 'RAG stands for Retrieval-Augmented Generation. It combines retrieval with generation.', 
                          'question': 'What is RAG'})
```

```python
# 4) create chain
prompt4 = ChatPromptTemplate.from_messages(
    [
        ('system', 'You are a helpful AI.'),
        ('human', "Summarize this text: {text}")
    ]
)

chain4 = prompt4 | model

result4 = chain4.invoke({"text" : """ LangChain is a framework for building applications with LLMs. It provides tools for prompts,
retrievers, chains, memory, and agents, making it easier to build production-ready systems."""})

print(result4.content)

chain4.invoke()
```

```python
# 5) RunnablePassthrough
from langchain_core.runnables import RunnablePassthrough, RunnableLambda, RunnableParallel

chain5 = RunnableParallel(
    original = RunnablePassthrough(),
    length = RunnableLambda(lambda x: len(x))
)

result5 = chain5.invoke('What is RAG?')
print(result5)
```

```python
# 6) Multi-input Prompt

prompt6 = ChatPromptTemplate.from_messages(
    [
    ('system', "Answer ONLY using the given context. If the answer isn't in the context, say 'Insufficient context."),
    ('human', 'Give answer to the following question: {question}, based on the context: {context}.')
    ]
)

chain6 = prompt6 | model
result6 = chain6.invoke(
    {
        'question' : 'How has AI impacted modern society?',
        "context" : """ Artificial Intelligence (AI) refers to the development of computer systems that can perform tasks typically requiring human intelligence, such as learning, reasoning, problem-solving, perception, and language understanding. The concept dates back to the 1950s, when researchers began exploring whether machines could simulate human thinking, but major advances have occurred in recent decades due to increased computing power, large datasets, and improved algorithms. Today, AI includes technologies like machine learning, deep learning, natural language processing, and computer vision, and it is widely used in applications such as virtual assistants, recommendation systems, healthcare diagnostics, finance, transportation, and automation, significantly influencing modern society and the global economy."""
    }
)

print(result6.content)
```

```python
# 7) 

def fake_retriever(query):
    return "Rag stands for Retrieval Augmented Generation"

retriever_runnable = RunnableLambda(fake_retriever)

prompt7 = ChatPromptTemplate.from_messages(
    [
        ('system', 'You are a helpful assistant.Use given context to answer'),
        ('human', 'Context: \n{context}\n\nQuestion:\n{question}')
    ]
)

chain7 = (RunnableParallel(
    context = retriever_runnable, 
    question = RunnablePassthrough()
)
| prompt7
| model
)

result7 = chain7.invoke("What is RAG?")
print(result7.content)
```

```python
# 8) 
prompt8 = ChatPromptTemplate.from_messages(
    [
        ('system',
         "You are a factual assistant. Answer ONLY using the provided context. "
         "If the answer is not in the context, respond exactly: Insufficient context."),
         ('human', 
          'Context:\n{context}\n\nQuestion:\n{question}')
    ]
)

chain8 = prompt8 | model
context = "RAG stands for Retrieval-Augmented Generation. It combines retrieval with generation."

res_ok = chain8.invoke({
    'context': context, 'question': 'What does RAG stand for?'
})
print('In-Context:', res_ok.content)

res_bad = chain8.invoke(
    {
        'context': context, 'question': "Who invented RAG an in what year?"
    }
)
print("Out-of-Context:", res_bad.content)
```
